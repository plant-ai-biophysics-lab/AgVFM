```python
from pathlib import Path

from vlme import run_gt_vs_predictions

# --- User inputs ---
NUM_IMAGES = 2
RANDOM_SEED = 100
IMAGES_DIR = Path("../_data/T4_REAL_FLOWER_SOBJ.v18-train_2023_test_location_genotype/test/images")
LABELS_DIR = Path("../_data/T4_REAL_FLOWER_SOBJ.v18-train_2023_test_location_genotype/test/labels")
MODEL_WEIGHTS_PATH = "model_weights/yolov8x-worldv2.pt"
GT_CLASS_NAMES = ["flower"]
OPEN_SET_CLASS_NAMES = [
    "small yellow flower", 
    "small white flower",
    ""
]
PREDICT_KWARGS = {
    "imgsz": 1280,
    "conf": 0.1,
    "max_det": 500,
    "verbose": False,
    "iou": 0.3
}

_, _, results_list = run_gt_vs_predictions(
    images_dir=IMAGES_DIR,
    labels_dir=LABELS_DIR,
    num_images=NUM_IMAGES,
    random_seed=RANDOM_SEED,
    class_names=GT_CLASS_NAMES,
    yolo_classes=OPEN_SET_CLASS_NAMES,
    weights_path=MODEL_WEIGHTS_PATH,
    predict_kwargs=PREDICT_KWARGS,
    ncols=2,
    show=True,
    return_results=True,
)
```
